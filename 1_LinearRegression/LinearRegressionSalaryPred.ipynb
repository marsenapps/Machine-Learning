# Linear Regression: <font color='red'>Salary Prediction</font>

## Import Library

```python
import pandas as pd # Data Manipulation and analysis
import numpy as np # multi-dimensional arrays and matrices and mathematical functions to operate on these arrays
import matplotlib.pyplot as plt # Plotting library
```

## Retrieve The Data

```python
# Load dataset
df = pd.read_csv("Salary_Data.csv")
```

```python
# Dispaly 5 rows
df.head()
```

```python
# Find important information on dataset
df.info()
```

Observation: 30 observations and non-null values

## Preparing the data

```python
## Select YearsExperience as Independent variable
# and the column of "Salary" as Dependent/target variable

X= df[["YearsExperience"]]
y=df["Salary"]
```

```python
## Put the points on the graph using "Scatter Plot"
plt.scatter(X,y)
plt.show()
```

### Spliting Training/Testing Dataset

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.3, random_state=0)
```

```python
# !pip install scikit-learn
```

```python
#import sklearn
#print("Version: ", sklearn.__version__)
```

## Build and train a Model

```python
# Import linearRegression library
from sklearn.linear_model import LinearRegression
lr =LinearRegression() # Instantiate
```

```python
lr.fit(X_train, y_train) # Trainin data set
```

### Getting Coefficients from Linear Equation

```python
# # $ y = \beta_0 + \beta_1 x + \epsilon$
```

$ y = \theta_0 + \theta_1 x + \epsilon $

```python
print("Coefficients: ",lr.coef_) # tetha-1
print("Intercept: ",lr.intercept_) #tetha-0, with y-intersection
```

### Add a line on the Graph

```python
coordindates = np.linspace(0,15,1000)
plt.scatter(X,y)
plt.plot(coordindates, lr.coef_[0]*coordindates + lr.intercept_, color='r', alpha=0.5)
plt.title("Experience vs Salary")
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

## Evaluate

Lets evaluate the previous model and checking its coefficients and interpret 

```python
# Prediction on the test dataset
y_pred = lr.predict(X_test)
```

#### Use All Regression Metrics

```python
from sklearn import metrics
print('MAE: ', metrics.mean_absolute_error(y_test, y_pred))
print('MSE: ', metrics.mean_squared_error(y_test, y_pred))
print('RMSE: ', np.sqrt(metrics.mean_squared_error(y_test, y_pred)))
print('R-2: ', metrics.r2_score(y_test, y_pred))
```

#### Let us use new Data with Experiences

```python
NewData = [[5],[4],[10]]
NewTest =pd.DataFrame(NewData,columns=['YearsExperience'])
y_new = lr.predict(NewTest)
print(y_new)
```
